A notebook to demonstate llm's ability to conduct a pattimandram. Pattimandram has total 5 llm's. 2 teams with 2 team members and a judge. They debate on a topic and the judge will score each members discussions and announce a final winner. 

LLMs
Judge - minimax/minimax-m3.1-flash:free

Team1
nvidia/nemotron-3.5-lightning:free
nvidia/nemotron-3.5-super-vl:free

Team2
nvidia/nemotron-3-ultra-550b-a55b:free - Provides standout response
#nvidia/nemotron-3-super-120b-a12b:free
deepseek/deepseek-v4.1-flash-1017 - Provides best responses or counter arguments

Step 1:

1. Install and import all the neccessary packages
2. Setup .env file
3. Intiaze the llms

Step 2:

1. System prompts for each llms
2. Functions to call each llms with system and user prompt

Step 3:

1. Final judge llm scores the debater llms and announce the winner.

```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display

load_dotenv(".env")

client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=os.environ["OPENROUTER_API_KEY"],
)

# response = client.chat.completions.create(
#     model="minimax/minimax-m3.1-flash:free",
#     messages=[{"role": "user", "content": "Introduce yourself briefly."}],
# )

# print(response.choices[0].message.content)
```

```python
raja_system = """
You are Raja from India, an expert debate speaker.

You have a unique speech style with a blend of everyday family humor, sharp wit, and deep literary insights. 
Unlike traditional scholarly speakers, you approaches the topic with the persona of an "ordinary observer", making your arguments highly relatable.
Your responses must stay focused on the topic.

Rules:
- Give witty and humorous response
- Speak like a expert debate speaker
- Keep the response short and interesting
- Response length must be between 50 and 100 words

You are part of Team 1, debating with Manikandan and Barathi and teaming with Gnanasambhantham. Your judge is Rangaraj.
"""

gnanasambhantham_system = """
You are Gnanasambhantham from India, a scholar and a researcher.

You respond with balanced but strong arguments backed by reasoning,
real-world examples, and technical insight.
You remain calm and intelligent while defending your viewpoint.

Rules:
- Deliver strong and engaging arguments
- Keep responses concise and focused
- Response length must be between 50 and 100 words
- Treat the opposition with the absolute dignity of a scholar, then gently dismantle their logic with pure sarcasm.
- An argument without a literary reference is just gossip; anchor your words in culture

You are part of Team 1, debating with Manikandan and Barathi and teaming with Raja. Your judge is Rangaraj.
"""

manikandan_system = """
You are Manikandan from India, a deeply rooted scholar and orator.

You embed rich storytelling from Hindu epics, Tamil scriptures, and Dharma into his points,
You utilizes situational irony, storytelling, and cultural anecdotes to generate clean, 
reflective humor that prompts audiences to "laugh and think simultaneously".

Rules:
- Give thoughtful and humorous responses
- Stay strictly on topic
- Response length must be between 50 and 100 words
- Keep your moral compass pointed toward traditional Dharma and Hindu heritage
- Let your wit serve as the sugar-coating for deep truth

You are part of Team 2, debating with Raja and Gnanasambhantham and teaming with Barathi. Your judge is Rangaraj.
"""

barathi_system = """
You are Barathi from India, you speak with scholarly depth, unmatched poise, and structured logic.

You argue with intense emotional conviction
You provide fact-based, insightful, and unconventional perspectives.
You are persuasive, intelligent, and capable of challenging both sides effectively.

Rules:
- Deliver strong and engaging arguments
- Let the opposition entertain the crowd with comedy; your job is to dismantle their logic with undeniable facts.
- Response length must be between 50 and 100 words
- Do not use bullet points
- Speak like an corporate professional

You are part of Team 2, debating with Raja and Gnanasambhantham and teaming with Manikandan. Your judge is Rangaraj.
"""

rangaraj_system = """
You are Rangaraj from India, a scholar, highly respected and impartial debate judge.
You evaluate arguments based on truthfulness, facts, clarity, logic, and persuasiveness.
Rules:
- Provide fair and unbiased evaluations
- Focus on the strength of arguments and delivery
- Keep evaluations concise and constructive

You are judging a debate between two teams - Team 1: Raja and Gnanasambhantham, and Team 2: Manikandan and Barathi.
"""
```

```python
conversation = """
Rangaraj: Welcome, everyone! Today’s debate is “Joint family vs Nuclear family” Team 1 will argue that Joint family gives more happiness, while Team 2 will argue that Nuclear families gives more happiness. Let’s begin with Team 1.

Team 1: You are the "Joint family" side. Make a persuasive, fact-based opening argument about how joint families benefits people. Address the audience and be ready to respond to Team 2.

Team 2: You are the “Nuclear family” side. Make a persuasive, fact-based opening argument about how nuclear families benefits people. Address the audience and be ready to challenge Team 1.

"""
```

```python
def call_raja(conversation):

    user_prompt = f"""
        You are Raja in a conversation with Gnanasambhantham, Manikandan and Barathi.

        Conversation so far:
        {conversation}

        Now respond as Raja.
    """

    response = client.chat.completions.create(
        model="nvidia/nemotron-3-ultra-550b-a55b:free",
        messages=[
            {"role": "system", "content": raja_system},
            {"role": "user", "content": user_prompt}
        ]
    )

    return response.choices[0].message.content
```

```python
def call_gnanasambhantham(conversation):

    user_prompt = f"""
        You are Gnanasambhantham in a conversation with Raja, Manikandan and Barathi.

        Conversation so far:
        {conversation}

        Now respond as Gnanasambhantham.
    """

    response = client.chat.completions.create(
        model="nvidia/nemotron-3.5-super-vl:free",
        messages=[
            {"role": "system", "content": gnanasambhantham_system},
            {"role": "user", "content": user_prompt}
        ]
    )
    
    return response.choices[0].message.content
```

```python
def call_manikandan(conversation):

    user_prompt = f"""
        You are Manikandan in a conversation with Raja, Gnanasambhantham and Barathi.

        Conversation so far:
        {conversation}

        Now respond as Manikandan.
    """

    response = client.chat.completions.create(
        model="nvidia/nemotron-3.5-lightning:free",
        messages=[
            {"role": "system", "content": manikandan_system},
            {"role": "user", "content": user_prompt}
        ]
    )
    
    return response.choices[0].message.content
```

```python
def call_barathi(conversation):

    user_prompt = f"""
        You are Barathi in a conversation with Raja, Gnanasambhantham and Manikandan.

        Conversation so far:
        {conversation}

        Now respond as Barathi.
    """

    response = client.chat.completions.create(
        model="deepseek/deepseek-v4.1-flash-1017",
        messages=[
            {"role": "system", "content": barathi_system},
            {"role": "user", "content": user_prompt}
        ]
    )
    
    return response.choices[0].message.content
```

```python
def call_rangaraj(conversation):
    
    user_prompt = f"""
        You are Rangaraj judging the debate between Team 1 - Raja, Gnanasambhantham and Team 2 -Manikandan and Barathi.

        Conversation so far:
        {conversation}

        Now announce the winner with a clear statement and justification.
    """

    response = client.chat.completions.create(
        model="minimax/minimax-m3.1-flash:free",
        messages=[
            {"role": "system", "content": rangaraj_system},
            {"role": "user", "content": user_prompt}
        ]
    )
    
    return response.choices[0].message.content
```

```python
display(Markdown("## Multi Model Conversation\n"))

display(Markdown(conversation.replace("\n", "  \n")))

for i in range(2):

    raja_reply = call_raja(conversation)
    conversation += f"\nRaja: {raja_reply}"

    display(Markdown(f"### Raja\n{raja_reply}\n"))

    manikandan_reply = call_manikandan(conversation)
    conversation += f"\nManikandan: {manikandan_reply}"

    display(Markdown(f"### Manikandan\n{manikandan_reply}\n"))

    gnanasambhantham_reply = call_gnanasambhantham(conversation)
    conversation += f"\nGnanasambhantham: {gnanasambhantham_reply}"

    display(Markdown(f"### Gnanasambhantham\n{gnanasambhantham_reply}\n"))

    barathi_reply = call_barathi(conversation)
    conversation += f"\nBarathi: {barathi_reply}"

    display(Markdown(f"### Barathi\n{barathi_reply}\n"))
```

```python
rangaraj_reply = call_rangaraj(conversation)
display(Markdown(f"### Rangaraj Judgement...\n{rangaraj_reply}\n"))
```
